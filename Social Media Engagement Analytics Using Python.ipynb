### Task 2 — Data Cleaning

#### cleaning missing values

```python
# finding missing values

df.isnull().sum()
```

```python
df.isna().sum()
```

```python
# filling missing values

num_cols = df.select_dtypes(include='number').columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns

for col in categorical_cols:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

### Task 6 — Data Visualization

#### MATPLOTLIB

```python
# SCATTER PLOT- LIKES VS IMPRESSIONS


plt.scatter(df['impression_count'], df['likes'])

plt.xlabel('Impressions')
plt.ylabel('Likes')
plt.title('Likes vs Impressions')

plt.show()
```

```python
# LINE CHART - DAILY ENGAGEMENT TRENDS

daily_engagement = df.groupby('posted_at')['engagement_rate'].mean()

plt.plot(daily_engagement.index, daily_engagement.values)

plt.xlabel('Date')
plt.ylabel('Average Engagement Rate')
plt.title('Daily Engagement Trend')

plt.show()
```

```python
# BAR CHART - POSTS BY CATEGORY


df['post_category'].value_counts().plot(kind='bar')

plt.xlabel('Post Category')
plt.ylabel('Number of Posts')
plt.title('Number of Posts by Category')

plt.show()
```

```python
# PIE CHART - GENDER DISTRIBUTION

gender_count = df['gender'].value_counts()

plt.pie(gender_count, labels=gender_count.index)

plt.title('Gender Distribution')

plt.show()
```

```python
# HISTOGRAM - AGE DISTRIBUTION


plt.hist(df['age'])

plt.xlabel('Age')
plt.ylabel('Number of Users')
plt.title('Age Distribution')

plt.show()
```

```python
# BOXPLOT - ENGAGEMENT RATE

plt.boxplot(df['engagement_rate'])

plt.ylabel('Engagement Rate')
plt.title('Engagement Rate Distribution')

plt.show()
```

#### SEABORN

```python
# Count Plot – Post Type

sns.countplot(data=df, x='post_type')

plt.title('Count of Post Types')
plt.xlabel('Post Type')
plt.ylabel('Count')

plt.show()
```

```python
 # Bar Plot – Average Likes by Category

sns.barplot(data=df, x='post_category', y='likes')

plt.title('Average Likes by Category')
plt.xlabel('Post Category')
plt.ylabel('Average Likes')

plt.xticks(rotation=45)
plt.show()
```

```python
 # Violin Plot – Followers vs Sentiment


sns.violinplot(data=df, x='sentiment', y='follower_count')

plt.title('Followers vs Sentiment')
plt.xlabel('Sentiment')
plt.ylabel('Follower Count')

plt.show()
```

```python
# Pair Plot – Numerical Features

sns.pairplot(df[['likes', 'comments', 'shares', 'engagement_rate']].dropna())

plt.show()
```
